```python
from langchain_openai import ChatOpenAI
from langchain.prompts import PromptTemplate
from langchain.tools import tool
from langgraph.prebuilt import create_react_agent

# Define the model
llm_model = ChatOpenAI(model="gpt-4o-mini")

# Define the system prompt
system_prompt = (
    "You are an AI agent that assists content creators in creating videos for different brands based on the brand type like clothing, footware, glasses, etc.\n"
    "If you find the keywords like clothing, footware or glasses, suggest scripts on the basis of that"
    "Give answers in a stepwise format and print each step on a different line."
)

# Define tools using @tool decorator with docstrings
@tool
def create_script(topic: str) -> str:
    """
    Create a script for a video based on the given topic.
    """
    return f"This is a script about {topic}"

@tool
def create_video(script: str) -> str:
    """
    Create a video using the provided script.
    """
    return f"Video created using script: {script}"

@tool
def create_voice_over(script: str) -> str:
    """
    Create a voice-over based on the provided script.
    """
    return f"Voice-over created for script: {script}"

# List of tools
tools = [create_script, create_video, create_voice_over]

# Create the agent
agent = create_react_agent(
    prompt=system_prompt,
    model=llm_model,
    tools=tools
)
```

```python
from langchain_core.messages.ai import AIMessage
from langchain_core.messages import HumanMessage

query = (
    "I have a brand with brand type **footwear**, brand name **aesha**, "
    "and I want to launch new products for summer. Suggest me video content ideas for that."
)

state = {
    "messages": [HumanMessage(content=query)]
}

#Invoke the agent
response = agent.invoke(state)

messages = response.get("messages", [])
ai_messages = [m.content for m in messages if isinstance(m, AIMessage)]

if ai_messages:
    print(ai_messages[-1])
else:
    print("No AI message in response.")
```
